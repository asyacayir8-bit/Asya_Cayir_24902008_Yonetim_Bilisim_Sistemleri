### 2. KÜTÜPHANELERİN VE MODÜLLERİN İÇE AKTARILMASI (IMPORT)

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Kendi yazdığımız modüller
from finans_modeli import Islem
from utils import menu_goster, sayi_kontrol, tarih_kontrol
from islem_yonetimi import gelir_ekle, gider_ekle, islemleri_listele, islem_sil
from dosya_islemleri import csv_kaydet, csv_oku
from analiz import verileri_dataframe_yap, toplam_gelir_gider, aylik_analiz, numpy_istatistik
from gorsellestirme import aylik_grafik, gelir_gider_bar, pasta_grafik
```

### 3. KONSOL UYGULAMASI VE ANA DÖNGÜ (CLI INTERFACE)

Aşağıdaki kod hücresinde projenin ana menü döngüsü (`main()`) yer almaktadır. Hücreyi çalıştırarak interaktif olarak menüyü kullanabilirsiniz.

**Önemli Not:** Menüden çıkmak için lütfen `7` (Çıkış) seçeneğini kullanınız. Bu işlem verilerinizi otomatik olarak `islemler.csv` dosyasına kaydedecektir.

```python
def main():
    dosya_adi = "islemler.csv"
    # Kayıtlı verileri CSV dosyasından yükle
    gelirler, giderler = csv_oku(dosya_adi)
    
    while True:
        menu_goster()
        secim = input("Yapmak istediğiniz işlemi seçiniz (1-8): ").strip()
        
        if secim == '1':
            gelir_ekle(gelirler, giderler)
        elif secim == '2':
            gider_ekle(giderler, gelirler)
        elif secim == '3':
            islemleri_listele(gelirler, giderler)
        elif secim == '4':
            df = verileri_dataframe_yap(gelirler, giderler)
            if df.empty:
                print("\n[Uyarı] Analiz yapabilmek için henüz hiçbir işlem girilmemiştir.")
            else:
                t_gelir, t_gider = toplam_gelir_gider(df)
                net = t_gelir - t_gider
                print("\n" + "="*40)
                print("         GENEL FİNANSAL ANALİZ")
                print("="*40)
                print(f"Toplam Gelir : {t_gelir:,.2f} TL")
                print(f"Toplam Gider : {t_gider:,.2f} TL")
                print(f"Net Durum    : {net:,.2f} TL ({'Kâr/Tasarruf' if net >= 0 else 'Zarar/Açık'})")
                print("-"*40)
                
                print("\n--- AYLIK BAZDA ÖZET ANALİZ ---")
                aylik = aylik_analiz(df)
                print(aylik)
                
                print("\n--- İSTATİSTİKSEL DETAYLAR (NumPy) ---")
                ist = numpy_istatistik(df)
                print(f"Ortalama İşlem Tutarı : {ist['ortalama']:,.2f} TL")
                print(f"Minimum İşlem Tutarı  : {ist['minimum']:,.2f} TL")
                print(f"Maksimum İşlem Tutarı : {ist['maksimum']:,.2f} TL")
                print(f"Standart Sapma        : {ist['standart_sapma']:,.2f} TL")
                print("="*40)
        elif secim == '5':
            df = verileri_dataframe_yap(gelirler, giderler)
            if df.empty:
                print("\n[Uyarı] Grafik çizmek için veri bulunmamaktadır.")
            else:
                # Grafikleri oluştur ve kaydet
                aylik_grafik(df, "aylik_analiz_cizgi.png")
                gelir_gider_bar(df, "gelir_gider_toplam_bar.png")
                pasta_grafik(df, "gelir_gider_pasta.png")
                print("\n[Sistem] Grafikler başarıyla oluşturuldu ve kaydedildi.")
                
                # Notebook üzerinde gösterim
                from IPython.display import Image, display
                print("\n--- Toplam Gelir/Gider Sütun Grafiği ---")
                display(Image(filename="gelir_gider_toplam_bar.png"))
                print("\n--- Gelir/Gider Oran Dağılım Pasta Grafiği ---")
                display(Image(filename="gelir_gider_pasta.png"))
                print("\n--- Aylık Trend Çizgi Grafiği ---")
                display(Image(filename="aylik_analiz_cizgi.png"))
        elif secim == '6':
            csv_kaydet(dosya_adi, gelirler, giderler)
        elif secim == '7':
            print("\n[Sistem] Verileriniz otomatik kaydediliyor...")
            csv_kaydet(dosya_adi, gelirler, giderler)
            print("Sistemden güvenli bir şekilde çıkış yapıldı. İyi günler dileriz!")
            break
        elif secim == '8':
            id_sil = input("Silmek istediğiniz işlemin ID'sini giriniz: ").strip()
            islem_sil(gelirler, giderler, id_sil)
        else:
            print("\nGeçersiz seçim! Lütfen 1-8 arasında bir menü seçeneği giriniz.")

# Programı çalıştırmak için aşağıdaki satırı yorumdan kaldırıp hücreyi çalıştırabilirsiniz:
# main()
```
